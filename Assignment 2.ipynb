```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "/content/image.jpg"

# Read RGB image
img_bgr = cv2.imread(image_path)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB).astype(np.float32)
R, G, B = img_rgb[:, :, 0], img_rgb[:, :, 1], img_rgb[:, :, 2]

# Case 1: Mean average
gray_case1 = ((R + G + B) / 3.0).astype(np.uint8)

# Case 2: Custom weights (R=0.7, G=0.2, B=0.1)
w_r, w_g, w_b = 0.7, 0.2, 0.1
gray_case2 = (w_r * R + w_g * G + w_b * B).astype(np.uint8)

# Display outputs
fig, axes = plt.subplots(1, 3, figsize=(14, 5))
axes[0].imshow(img_rgb.astype(np.uint8))
axes[0].set_title("Original RGB")
axes[0].axis('off')

axes[1].imshow(gray_case1, cmap='gray')
axes[1].set_title("Case 1: Mean Average")
axes[1].axis('off')

axes[2].imshow(gray_case2, cmap='gray')
axes[2].set_title(f"Case 2: Weighted ({w_r}, {w_g}, {w_b})")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
