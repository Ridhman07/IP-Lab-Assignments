```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Replace with your actual uploaded image file path
image_path = "/content/image.jpg"

# Run the processing (Example: Assignment 1)
img_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Case 1: Mean threshold
mean_thresh = np.mean(img_gray)
binary_case1 = np.where(img_gray >= mean_thresh, 255, 0).astype(np.uint8)

# Case 2: Custom threshold
user_thresh = 128
binary_case2 = np.where(img_gray >= user_thresh, 255, 0).astype(np.uint8)

# Display outputs
fig, axes = plt.subplots(1, 3, figsize=(12, 4))
axes[0].imshow(img_gray, cmap='gray')
axes[0].set_title("Original")
axes[0].axis('off')

axes[1].imshow(binary_case1, cmap='gray')
axes[1].set_title(f"Mean Thresh ({mean_thresh:.1f})")
axes[1].axis('off')

axes[2].imshow(binary_case2, cmap='gray')
axes[2].set_title(f"Custom Thresh ({user_thresh})")
axes[2].axis('off')

plt.tight_layout()
plt.show()
```
