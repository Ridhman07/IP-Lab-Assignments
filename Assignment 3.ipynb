```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image_path = "/content/image.jpg"

# Load grayscale and generate binary image
img_gray = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
_, img_binary = cv2.threshold(img_gray, 127, 255, cv2.THRESH_BINARY)

# Parameters
border_width = 20
border_color_gray = 255  # White border for grayscale
border_color_bin = 255   # White border for binary

# 1. Border Padding
pad_gray = cv2.copyMakeBorder(img_gray, border_width, border_width, border_width, border_width,
                              cv2.BORDER_CONSTANT, value=border_color_gray)
pad_bin = cv2.copyMakeBorder(img_binary, border_width, border_width, border_width, border_width,
                             cv2.BORDER_CONSTANT, value=border_color_bin)

# 2. Image Complement
comp_gray = 255 - img_gray
comp_bin = cv2.bitwise_not(img_binary)

# Display outputs
fig, axes = plt.subplots(2, 3, figsize=(14, 9))

axes[0, 0].imshow(img_gray, cmap='gray')
axes[0, 0].set_title("Original Grayscale")
axes[0, 0].axis('off')

axes[0, 1].imshow(pad_gray, cmap='gray')
axes[0, 1].set_title(f"Padded Grayscale (w={border_width})")
axes[0, 1].axis('off')

axes[0, 2].imshow(comp_gray, cmap='gray')
axes[0, 2].set_title("Grayscale Complement")
axes[0, 2].axis('off')

axes[1, 0].imshow(img_binary, cmap='gray')
axes[1, 0].set_title("Original Binary")
axes[1, 0].axis('off')

axes[1, 1].imshow(pad_bin, cmap='gray')
axes[1, 1].set_title(f"Padded Binary (w={border_width})")
axes[1, 1].axis('off')

axes[1, 2].imshow(comp_bin, cmap='gray')
axes[1, 2].set_title("Binary Complement")
axes[1, 2].axis('off')

plt.tight_layout()
plt.show()
```
